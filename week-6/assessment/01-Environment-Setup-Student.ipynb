## Troubleshooting Section

### Common Issues and Solutions:

**1. File Upload Issues:**
- Ensure CSV files are uploaded to the correct Databricks workspace location
- Check file names match exactly: `banking_transactions.csv`, `customer_profiles.csv`, `fraud_rules.json`
- Verify file permissions and accessibility

**2. Schema Issues:**
- If inferSchema fails, manually define schema
- Check for special characters in column names
- Verify date formats in CSV files

**3. Memory Issues:**
- If working with larger datasets, consider using `.cache()` on DataFrames
- Adjust Spark configuration if needed

**4. JSON Parsing Issues:**
- Each line in fraud_rules.json should be a valid JSON object
- Check for proper JSON formatting

**5. Performance Tips:**
- Use `.cache()` on frequently accessed DataFrames
- Consider partitioning large datasets
- Use appropriate Spark cluster size for your data volume
